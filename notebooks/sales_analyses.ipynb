```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dir_path = Path.cwd().parent

data_path = dir_path / "data" / "sales_data.csv"
```

```python
database = pd.read_csv(data_path)
```

```python
database.isnull().sum()
```

```python
database.columns
```

```python
database.drop(columns=['Unnamed: 21', 'Unnamed: 22',
       'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25', 'Unnamed: 26',
       'Unnamed: 27'], inplace=True)
```

```python
database.isnull().sum()
```

```python
database.head()
```

```python
database.info()
```

```python
database['data'] = pd.to_datetime(database['data'])
```

```python
database[['preco_unitario','pct_desconto','gross_revenue','discount_value','net_revenue','cost','profit','margin_pct']]
```

```python
# Colunas que têm "R$" e caracteres especiais

colunas_moeda = ['preco_unitario', 'gross_revenue', 'discount_value', 'net_revenue', 'cost', 'profit']

for col in colunas_moeda:
    database[col] = database[col].str.replace('R$', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100

# Colunas de porcentagem (tratadas individualmente por terem regras diferentes)

database['pct_desconto'] = database['pct_desconto'].str.replace('%', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100
database['margin_pct'] = database['margin_pct'].str.replace(',', '', regex=False).astype(float) / 100
```

```python
database.info()
```

```python
traducao_colunas = {
    'customer_segment': 'segmento_cliente',
    'gross_revenue': 'receita_bruta',
    'discount_value': 'valor_desconto',
    'net_revenue': 'receita_liquida',
    'cost': 'custo',
    'profit': 'lucro',
    'margin_pct': 'margem_pct'
}

database.rename(columns=traducao_colunas, inplace=True)
```
