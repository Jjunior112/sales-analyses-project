```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dir_path = Path.cwd().parent

data_path = dir_path / "data" / "sales_data.csv"
```

```python
database = pd.read_csv(data_path)
```

```python
database.isnull().sum()
```

```python
database.columns
```

```python
database.drop(columns=['Unnamed: 21', 'Unnamed: 22',
       'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25', 'Unnamed: 26',
       'Unnamed: 27'], inplace=True)
```

```python
database.isnull().sum()
```

```python
database.head()
```

```python
database.info()
```

```python
database['data'] = pd.to_datetime(database['data'])
```

```python
meses = {
    'Janeiro': 1,
    'Fevereiro': 2,
    'Março': 3,
    'Abril': 4,
    'Maio': 5,
    'Junho': 6,
    'Julho': 7,
    'Agosto': 8,
    'Setembro': 9,
    'Outubro': 10,
    'Novembro': 11,
    'Dezembro': 12
}

database['mes_num'] = database['mês'].map(meses)
```

```python
database[['preco_unitario','pct_desconto','gross_revenue','discount_value','net_revenue','cost','profit','margin_pct']]
```

```python
# Colunas que têm "R$" e caracteres especiais

colunas_moeda = ['preco_unitario', 'gross_revenue', 'discount_value', 'net_revenue', 'cost', 'profit']

for col in colunas_moeda:
    database[col] = database[col].str.replace('R$', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100

# Colunas de porcentagem (tratadas individualmente por terem regras diferentes)

database['pct_desconto'] = database['pct_desconto'].str.replace('%', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100
database['margin_pct'] = database['margin_pct'].str.replace(',', '', regex=False).astype(float) / 100
```

```python
database.info()
```

```python
traducao_colunas = {
    'customer_segment': 'segmento_cliente',
    'gross_revenue': 'receita_bruta',
    'discount_value': 'valor_desconto',
    'net_revenue': 'receita_liquida',
    'cost': 'custo',
    'profit': 'lucro',
    'margin_pct': 'margem_pct'
}

database.rename(columns=traducao_colunas, inplace=True)
```

```python
database.info()
```

```python
#1 - Faturamento total

import locale

faturamento_total = database['receita_liquida'].sum()

locale.setlocale(locale.LC_ALL, 'pt_BR.UTF-8')
print(f'{locale.currency(faturamento_total, grouping=True)}')
```

```python
# 1.1 - Faturamento por mês

faturamento_mes = database.groupby(['ano','mês','mes_num'])['receita_liquida'].sum().reset_index().sort_values(by=['ano','mes_num'])

faturamento_mes['receita_liquida'] = faturamento_mes['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_mes
```

```python
# 2 - Lucro total

import locale

lucro_total = database['lucro'].sum()

locale.setlocale(locale.LC_ALL, 'pt_BR.UTF-8')
print(f'{locale.currency(lucro_total, grouping=True)}')
```

```python
# 2.1 - Lucro por mês

lucro_mes = database.groupby(['ano','mês','mes_num'])['lucro'].sum().reset_index().sort_values(by=['ano','mes_num'])

lucro_mes['lucro'] = lucro_mes['lucro'].apply(lambda x: locale.currency(x, grouping=True))

lucro_mes
```

```python
# 3 - Faturamento por região

faturamento_total_regiao = database.groupby(['regiao'])['receita_liquida'].sum().reset_index().sort_values(by=['regiao'])

faturamento_total_regiao['receita_liquida'] = faturamento_total_regiao['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_total_regiao
```

```python
# 3.1 - Faturamento por região e ano

faturamento_regiao = database.groupby(['ano','regiao'])['receita_liquida'].sum().reset_index().sort_values(by=['ano','regiao'])

faturamento_regiao['receita_liquida'] = faturamento_regiao['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_regiao
```

```python
# 3.2 lucro total por região
 
lucro_regiao = database.groupby(['regiao'])['lucro'].sum().reset_index().sort_values(by=['regiao'])

lucro_regiao['lucro'] = lucro_regiao['lucro'].apply(lambda x: locale.currency(x, grouping=True))

lucro_regiao 
```

```python
# 3.3 lucro total por região e ano
 
lucro_regiao = database.groupby(['ano','regiao'])['lucro'].sum().reset_index().sort_values(by=['ano','regiao'])

lucro_regiao['lucro'] = lucro_regiao['lucro'].apply(lambda x: locale.currency(x, grouping=True))

lucro_regiao 
```

```python
#faturamento por produto e ano

faturamento_produto = database.groupby(['produto_nome','ano'])['receita_liquida'].sum().reset_index().sort_values(by=['produto_nome','ano'])

faturamento_produto['receita_liquida'] = faturamento_produto['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_produto
```

```python
#faturamento por segmento

faturamento_segmento = database.groupby(['segmento_cliente','ano'])['receita_liquida'].sum().reset_index().sort_values(by=['segmento_cliente','ano'])

faturamento_segmento['receita_liquida'] = faturamento_segmento['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_segmento
```

```python
lucro_segmento = database.groupby(['segmento_cliente','ano'])['lucro'].sum().reset_index().sort_values(by=['segmento_cliente','ano'])

lucro_segmento['lucro'] = lucro_segmento['lucro'].apply(lambda x: locale.currency(x, grouping=True))

lucro_segmento
```

```python
#faturamento por canal de venda

faturamento_canal_venda = database.groupby(['canal_venda','segmento_cliente'])['receita_liquida'].sum().reset_index().sort_values(by=['canal_venda'])

faturamento_canal_venda['receita_liquida'] = faturamento_canal_venda['receita_liquida'].apply(lambda x: locale.currency(x, grouping=True))

faturamento_canal_venda
```
