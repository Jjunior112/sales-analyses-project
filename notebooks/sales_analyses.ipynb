```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dir_path = Path.cwd().parent

data_path = dir_path / "data" / "sales_data.csv"
```

```python
database = pd.read_csv(data_path)
```

```python
database.isnull().sum()
```

```python
database.columns
```

```python
database.drop(columns=['Unnamed: 21', 'Unnamed: 22',
       'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25', 'Unnamed: 26',
       'Unnamed: 27'], inplace=True)
```

```python
database.isnull().sum()
```

```python
database.head()
```

```python
database.info()
```

```python
database['data'] = pd.to_datetime(database['data'])
```

```python
database[['preco_unitario','pct_desconto','gross_revenue','discount_value','net_revenue','cost','profit','margin_pct']]
```

```python
# Colunas que têm "R$" e caracteres especiais

colunas_moeda = ['preco_unitario', 'gross_revenue', 'discount_value', 'net_revenue', 'cost', 'profit']

for col in colunas_moeda:
    database[col] = database[col].str.replace('R$', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100

# Colunas de porcentagem (tratadas individualmente por terem regras diferentes)

database['pct_desconto'] = database['pct_desconto'].str.replace('%', '', regex=False).str.replace(',', '', regex=False).str.replace('.', '', regex=False).astype(float) / 100
database['margin_pct'] = database['margin_pct'].str.replace(',', '', regex=False).astype(float) / 100
```

```python
database.info()
```

```python
traducao_colunas = {
    'customer_segment': 'segmento_cliente',
    'gross_revenue': 'receita_bruta',
    'discount_value': 'valor_desconto',
    'net_revenue': 'receita_liquida',
    'cost': 'custo',
    'profit': 'lucro',
    'margin_pct': 'margem_pct'
}

database.rename(columns=traducao_colunas, inplace=True)
```

```python
#1 - Qual faturamento total, por região e por estado?

faturamento_total  = database['receita_liquida'].sum()
faturamento_regiao = database.groupby(['regiao'])['receita_liquida'].sum().sort_values(ascending=False)
faturamento_estado = database.groupby(['estado'])['receita_liquida'].sum().sort_values(ascending=False)
```

```python
import locale

locale.setlocale(locale.LC_ALL, 'pt_BR.UTF-8')


sns.set_style('whitegrid')
sns.set_palette('Set2')
plt.figure(figsize=(10, 6))
plt.title('Faturamento Total por Região', fontsize=16)


ax = sns.barplot(x=faturamento_regiao.index, y=faturamento_regiao.values)

for container in ax.containers:
    # O locale.currency aplica o R$, os pontos e a vírgula perfeitamente
    labels = [locale.currency(v, grouping=True) for v in container.datavalues]
    ax.bar_label(container, padding=3, labels=labels)

plt.margins(y=0.1)

ax.set_yticklabels([]) 
ax.set_xlabel('')
plt.xlabel('Região', fontsize=14)

ax.grid(False)

plt.show()
```

```python
import locale

locale.setlocale(locale.LC_ALL, 'pt_BR.UTF-8')


sns.set_style('whitegrid')
sns.set_palette('Set2')
plt.figure(figsize=(23, 8))
plt.title('Faturamento Total por Estado', fontsize=16)


ax = sns.barplot(x=faturamento_estado.index, y=faturamento_estado.values)

for container in ax.containers:
    # O locale.currency aplica o R$, os pontos e a vírgula perfeitamente
    labels = [locale.currency(v, grouping=True) for v in container.datavalues]
    ax.bar_label(container, padding=3, labels=labels)

plt.margins(y=0.1)

ax.set_yticklabels([]) 
ax.set_xlabel('')
plt.xlabel('Estado', fontsize=14)

ax.grid(False)

plt.show()
```
